2. Con la reducción del IGV para restaurantes y hoteles (https://emprender.sunat.gob.pe/tributando/temas-actualidad/reduccion-igv-para-restaurantes-hoteles), los restaurantes pagan el 10% de IGV en vez del 18%. Un restaurante factura 3426 soles diarios, en promedio, después de impuestos. Calcule el valor base (el IGV es 10% sobre el valor base, ó 1 + .1).

Calcule, utilizando el valor base, cuánto es el descuento del IGV en soles (o el 8% de este valor base).

Cree una variable para cada uno de los siguientes:
- porcentaje de IGV 
- el total de facturación
- el monto del impuesto
- el valor base sin impuesto
- el descuento de IGV en soles (18% - 10%). Puedes comprobar tu fórmula con cualquier recibo de pago que detalle el IGV.

````python
porcentaje_igv_reducido = 0.10  
facturacion_diaria = 3426  #Facturación promedio después de impuestos

#Cálculo del valor base sin impuestos
valor_base = facturacion_diaria / (1 + porcentaje_igv_reducido)

#Monto del impuesto con IGV reducido
monto_impuesto_reducido = valor_base * porcentaje_igv_reducido

#Descuento del IGV en soles (18% - 10%)(8% de diferencia entre IGV del 18% y el 10%)
descuento_igv = valor_base * 0.08 

#Respuestas
print("Porcentaje de IGV reducido: {:.1f}%".format(porcentaje_igv_reducido * 100))
print(f"Total de facturación (después de impuestos): {facturacion_diaria} soles")
print(f"Valor base sin impuestos: {valor_base:.2f} soles")
print("Monto del impuesto reducido (10%): {:.2f} soles".format(monto_impuesto_reducido))
print("Descuento del IGV (18% - 10%): {:.2f} soles".format(descuento_igv))
````

Para ello tomará dos argumentos: la temperatura a convertir (float) y si está en Fahrenheit (F) o Celsius (C) (hint: puedes usar un string "F" o "C"). 

Si la temperatura está en F, la fórmula es:        
$F = 1.8*C + 32$    
Si la temperatura está en C, la fórmula es:   
$C = (F - 32)/1.8 $

Puedes llamar a tu función como gustes, el código comienza como `transforma_temperatura`

````python

def transforma_temperatura(temperatura, escala):
    """
    Convierte temperaturas entre Celsius y Fahrenheit.
    
    Args:
        temperatura (float): La temperatura a convertir.
        escala (str): La escala de la temperatura. "C" para Celsius, "F" para Fahrenheit.
    
    Returns:
        str: La temperatura convertida con su unidad correspondiente o un mensaje de error si la escala no es válida.
    """
    if escala == "C": #(Convertir de Celsius a Fahrenheit)
        resultado = 1.8 * temperatura + 32
        return f"{resultado:.2f}°F"
    elif escala == "F":  #(Convertir de Fahrenheit a Celsius)
        resultado = (temperatura - 32) / 1.8
        return f"{resultado:.2f}°C"
    else:
        return "Escala no válida. Usa 'C' para Celsius o 'F' para Fahrenheit."

print(transforma_temperatura(0, "C"))   
print(transforma_temperatura(32, "F"))  
print(transforma_temperatura(100, "C"))
print(transforma_temperatura(212, "F")) 
````
